```python
import torch
import torch.nn as nn
```

```python
class RNNModule(nn.Module):
  def __init__(self):
    super(RNNModule, self).__init__()
    self.rnn = nn.RNN(input_size=5,hidden_size=3,batch_frist=True)
    self.fc = nn.Linear(3,1)
    self.sigmoid = nn.Sigmoid()



#forward

def forward(self,x):
  out,_ = self.rnn(x)
  out = out(out[:,-1,:])
  out = self.fc(out)
  out = self.sigmoid(out)
```
