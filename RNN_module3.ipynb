**Tokenize**

```python
import re
def tokenize(text):
  #Lowercase
  text = text.lower()


  #romoved quotes
  text = re.sub(r"[\"']","",text)
  text = re.sub(r"[‘’]","",text)

  #removed all punctuation
  text = re.sub(r"[^\w\s]","",text)
  punctuations = re.sub(r"[^a-z0-9\s]"," ",text)

  #removed extra space
  text = text.replace("\n", " ")


  #tokenize split into words
  tokens = text.split()

  return tokens
```

vocab formating

```python
vocab ={'<UNK>':0}
```

```python
vocab
```

```python
def build_vocab(row):
 # print(row['question'],row['answer'])
  tokenize_question = tokenize(row['question'])
  tokenize_answer = tokenize(row['answer'])


  #print(tokenize_question)
  #print(tokenize_answer)

  # merge tokenize
  merge_tokens = tokenize_question + tokenize_answer

  #print(merge_tokens)

  for token in merge_tokens:
    if token not in vocab:
      vocab[token] = len(vocab)

```

```python
df.apply(build_vocab,axis=1)
```

```python
vocab
```

**text** **to** **indices**

```python
def text_to_indices(text,vocab):
  indexed_text = []

  for token in tokenize(text):
    #print(token)
    if token in vocab:
      indexed_text.append(vocab[token])
    else:
      indexed_text.append(vocab['<UNK>'])

  return indexed_text
```

# Dataset and dataloader

```python
import torch
from torch.utils.data import Dataset,DataLoader
```

```python
class QADataset(Dataset):
  def __init__(self,df,vocab):

    self.df = df
    self.vocab = vocab

  def __len__(self):
    return self.df.shape[0]

  def __getitem__(self,idx):
    numerical_question = text_to_indices(self.df.iloc[index]['question'],vocab)
    numerical_answer = text_to_indices(self.df.iloc[index]['answer'],vocab)


    return torch.tensor(numerical_question),torch.tensor(numerical_answer)
```

```python
dataset = QADataset(df,vocab)
```

# DataLoader

```python
dataloader = DataLoader(dataset,batch_size=1,shuffle=True)
```

# RNN Architecture

```python
import torch.nn as nn

class simpleRNN(nn.Module):
  def __init__(self, vocab_size, embedding_dim = 50 , hidden_size=64):
    super(simpleRNN, self).__init__()

    self.embedding = nn.Embedding(vocab_size, embedding_dim)
    self.rnn = nn.RNN(embedding_dim, hidden_size, batch_first=True)
    self.fc = nn.Linear(hidden_size, vocab_size)

# forward
  def forward(self, x):
    x = self.embedding(x) # x: (batch_size, seq_len, embedding_dim)
    _, final = self.rnn(x) # final: (num_layers * num_directions, batch_size, hidden_size)

    # Squeeze the first dimension of 'final' to get (batch_size, hidden_size)
    final_hidden_state = final.squeeze(0)

    # Pass the final hidden state to the linear layer
    x = self.fc(final_hidden_state) # x: (batch_size, vocab_size)
    return x
```

```python
model = simpleRNN(len(vocab))
```

```python
model
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
epochs = 50
```

```python
for epoch in range(epochs):
  total_loss = 0
  for question,answer in dataloader:
    optimizer.zero_grad()
    output = model(question)

    target = answer[0][0].unsqueeze(0)

    loss = criterion(output, target)
    loss.backward()
    optimizer.step()
    total_loss += loss.item()

  if (epoch+1)%10 ==0:
    print(f"Epoch{epoch+1:3d}/{epoch} | Loss : {total_loss : .4f}")
```

# Prediction

```python
idx_to_word = {index: word for word, index in vocab.items()}
```

```python
idx_to_word
```

```python
def prediction(model, question, idx_to_word):
  model.eval()
  with torch.no_grad():
    indices = text_to_indices(question, vocab)
    if not indices:
      return "<UNK>"

    x = torch.tensor(indices).unsqueeze(0)

    pred = torch.argmax(model(x), dim=1).item()

    return idx_to_word[pred]
```
